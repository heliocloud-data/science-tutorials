## SuperMAG 1-second data

A demo accessing the SuperMAG 1-second data.  Each file is a DataMap containing all the station data available for that time.  Our demo first uses CloudCatalog to fetch the list of datafiles for a given time range.  We then, for the first 3 files, either (1) plot a heatmap of all stations or (2) plot the vector timeseries for all stations at the timestep within the lat/lon bounding box of lat 40:60, lon -5:5 (which generally is ~3 stations).

This demo requires two Python helper function sets, 'PyDMAP.py' and 'supermag_tools.py', provided. These will soon be available as a 'pip install' but for now are distributed manually with this demo.

```python
import cloudcatalog
import supermag_tools as sm
import PyDMAP as dmap

dataset_id1 = 'supermag_mag-01s-global'
start = '2020-02-01T00:00:00Z'
stop =   '2020-02-02T00:00:00Z'
fr = cloudcatalog.CloudCatalog("s3://gov-nasa-hdrl-data1/")
filekeys_id1 = fr.request_cloud_catalog(dataset_id1,start_date=start,stop_date=stop)
```

```python
mychoice = int(input("1 for heatmap (default), 2 for individual stations in box: (1/2) "))
for i in range(3):
    s3uri = filekeys_id1.iloc[i]['datakey']
    mydmap, err = sm.stream_dmap_from_s3(s3uri)
    if mychoice == 2:
        ids, indices = sm.get_stations_in_box(mydmap, 40, 60, -5, 5)
        print(f"Operation on station set {ids}")
        for station in ids:
            sm.plot_station_timeseries(mydmap, station)
    else:
        sm.all_stations_heatmap(mydmap)
```

```python
# Also a stub showing how to run a lambda calculation on a set of S3 files fetched by CloudCatalog!
print("File hash, start_time, end_time, filesize")
fr.stream(filekeys_id1[0:10], lambda bo, d, e, f: print(hash(bo.read()), d.replace(' ', 'T')+'Z', e.replace(' ','T')+'Z',f))
```
