# Getting Started with Jupyter AI

The following is a tutorial on getting started with the Jupyter AI extension. Information in this notebook is modified from Jupyter AI's documentation at https://jupyter-ai.readthedocs.io/en/v3/index.html (BSD 3-Clause License).

Welcome to the Jupyter AI integration with HelioCloud! Click on the chat icon to the left, and you will open the Jupyter AI extension panel. You can click on the plus (+) icon in the top left corner of the panel to start a new chat.

## AI Personas

In the chat message field at the bottom of the panel, type "@" to see the agents that are available to you. Currently, HelioCloud has support for Claude, Codex, Gemini, and Copilot.

## Using the Chat

You can chat with specific agents by appending @ to the agent name, and following it with a prompt. For example ```@Codex How do I print "hello world" in python?"```. If it is your first time using an agent in the Jupyter AI chat, the agent will not respond. Instead, it will prompt you to log in via the terminal. 

The Jupyter MCP server allows you to run shell commands, read files, and interact with notebooks from the chat panel. It will always ask permission to perform any action other than reading the files within the current workspace.
