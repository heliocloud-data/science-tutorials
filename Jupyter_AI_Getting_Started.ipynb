# Getting started with Jupyter AI

HelioCloud now inlcudes the Jupyter AI extension (https://github.com/jupyterlab/jupyter-ai), allowing you to accelerate your workflow with the help of generative AI. For now, the extension will only support cloud-based models from OpenAI, Anthropic, and Google. These models require an API key, which you can get through a personal subscription, or through your employer's enterprise license. If your employer does not have an enterprise license with any of these providers, or you do not wish to pay for an API key out of pocket, then you can easily obtain a free API key from many different sources.
