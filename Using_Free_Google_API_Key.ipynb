# Using a Free Google AI Studio API Key

In this notebook, we will show you how to leverage Google's free API key for use with the Jupyter AI extension within this HelioCloud instance. First, you must obtain the API key from Google AI Studio. 

## Steps to obtain a free Google API key

1. Go to https://aistudio.google.com/app/api-keys
2. After logging in and accepting the terms of service, click on "Get API key" in the bottom left corner of the page
3. On the next page, click "Create API key" on the top right of the page
4. Give your key a new name, or leave it with the default name
5. Under "Choose an imported project," you can either import an existing project from your Google Cloud, or create a new project. If you don't have an existing project to import, click "+ Create project" and name the new project.
6. One you have named your key, and selected your project, click "Create key."
7. You will see your key listed in the table now. You can click on the key in the "Key" column and copy your new API key from here. You can also see in the "Quota tier" column, it should say "Free tier," under "Set up billing." You do not need to set up billing unless you would like to switch from the free tier to a paid tier.

## How to use the API key

Now that you have secured your free API key from Google AI Studio, you can use it with the Jupyter AI extension. If you have read through the "Jupyter_AI_Getting_Started.ipynb" notebook, then you know that there are two ways to use the extension: using the chat UI or with in-notebook ```%%ai``` magic.

### The chat UI 

If you would like to set up your API key for the chat UI, you will need to click on the Jupyter AI Chat icon on the left hand navigation bar. On the top right of the chat window is a gear icon--click on it to access the Jupyter AI model settings. At the top, under the words "Language model" is a dropdown menu. Select "Gemini :: gemini-2.5-flash." We are chosing this model out of all of the others because this is the only one that is covered by the free Google API key. Be warned though, this model is strictly limited by google to only 20 requests PER DAY. That means that the chat will stop working after you have sent 20 messages. 

A limit of 20 requests per day certainly seems meager. Other models offer 1,000 requests per day, or more. This is the unfortunate trade-off of a free API key. Not to worry, though. We will implement a slightly different model setup for the Jupyter AI magics, which include the ```%ai``` line magic and the ```%%ai``` cell magic. The magic commands are where Jupyter AI really shines, and where you, as the user, will get the most out of it.

### Jupyter AI Magics

Fortunately, there is another model that is accessible through Google's free API key that allows for 14,400 requests per day. This model is called Gemma 3 27B, and it is what we will use for ```%ai``` line magic and ```%%ai``` cell magic. While it is not as high-performing as a Gemini model, it is still an excellent model and is well suited for the highly iterative work required for developing in a Jupyter notebook. 

In order to use Jupyter AI magics, we must set our API key as an environmental variable. To keep your API key secure, we will use the ```getpass()``` method. Go ahead and run the cell below. Just remember that in order to paste things into a Jupyter notebook, must use ```shift + right-click``` and then select paste from the context menu.

````python
import os
import getpass

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your GOOGLE API key: ")
````

Currently, Jupyter AI does not support Gemma models out of the box. However, Google exposes Gemma 3 27B through an OpenAI compatible API, which Jupyter AI *does* support. All we have to do is point Jupyter AI to the OpenAI endpoint where Google models are hosted. We will create a new environmental variable ```OPENAI_BASE_URL``` and set it equal to the endpoint at "https:\/\/generativelanguage.googleapis.com\/v1beta\/openai\/." Since Jupyter expects an OpenAI API key for use with this url, we will create a new variable called ```OPENAI_API_KEY``` and set it equal to ```GOOGLE_API_KEY```. 

Run the cell below, and we are nearly ready to start using magic commands.

````python
os.environ["OPENAI_API_KEY"] = os.environ["GOOGLE_API_KEY"]
os.environ["OPENAI_BASE_URL"] = "https://generativelanguage.googleapis.com/v1beta/openai/"
````

## Load Jupyter AI and call the model

Now we load Jupyter AI

````python
%load_ext jupyter_ai
````

Now we invoke the the model using Jupyter AI's ```<provider-id>:<local-model-id>``` syntax. Since Google's OpenAI-compatible endpoint implements OpenAI's chat API, we set the ```<provider-id>``` to ```openai-chat``` and the ```<local-model-id>``` to ```gemma-3-27b-it```. For the curious, a breakdown of the model id ```gemma-3-27b-it``` is as follows:
- ```gemma```: Google's Gemma model family
- ```3```: 3rd generation
- ```27b```: The size of the neural network, in this case, it uses 27 billion parameters
- ```it```: Instruction-tuned, meaning it is trained to follow human prompts

````python
%%ai openai-chat:gemma-3-27b-it
Hello - just reply with the sentence "Hi! I'm Gemma!"
````

If you find that the model response time is far too slow for your liking, you can always try using one of the following smaller Gemma models:
- ```gemma-3-12b-it```
- ```gemma-3-4b-it```
- ```gemma-3-1b-it```
Smaller models are often (but not always) faster. Just keep in mind that, in general, the fewer parameters a model has, the weaker its reasoning.

Assuming you haven't already hit your daily token limit for a Gemini model in the chat UI, you can also try invoking a Gemini model for cell magic.

````python
%%ai gemini:gemini-2.5-flash
Hello — just reply with the sentence "Hi! I'm Gemini!"
````

## Comparing Gemma and Gemini

When you run the models with the same prompt below, you'll probably find that they provide similar solutions and even similar reasoning for those solutions. This is usually the case among strong models for common scientific coding tasks. 
However, when compared to Gemma's response, Gemini's output will likely contain:
- Tutorial-style language
- Longer and more structured explanations
- Well-managed edge cases and added safeguards
- More detailed examples
- Higher-level statistical choices

Fortunately, Gemini's advantages over Gemma don't discount the fact that Gemma 3 27B is still a very capable model. You can easily tweak the prompt to obtain more Gemini-like responses from Gemma as well. 

## Comparing Gemma and Gemini

Gemini and Gemma are both strong models, but there is a reason Gemini is limited to **20 requests/day** while Gemma can handle up to **14,400 requests/day**. When you run the models with the same prompt below, you will likely find that they produce similar solutions and sometimes even similar reasoning. This is common among strong models when solving standard scientific coding tasks.

However, compared to Gemma’s response, Gemini’s output will often include:
- Tutorial-style language
- Longer, more structured explanations
- More careful handling of edge cases and safeguards
- More detailed examples
- Slightly higher-level statistical choices

Gemma’s responses are still , on the other hand, often emphasize clarity and direct implementation. Its code is typically shorter and more transparent, making it easier to quickly understand, modify, and integrate into an existing research workflow.

These differences reflect different design priorities: Gemini tends to emphasize robustness and explanation, while Gemma focuses on efficiency, transparency, and high-throughput use cases.

Overall, Gemini tends to emphasize robustness and explanation, while Gemma provides efficiency and simpler, more read use cases.

Importantly, Gemini’s advantages do not diminish the fact that Gemma 3 27B is still a very capable model. In fact, small prompt adjustments can often encourage Gemma to produce responses that are closer in style to Gemini’s. 

````python
%%ai openai-chat:gemma-3-27b-it

You have a NumPy array containing noisy experimental measurements.
Show a concise Python example for estimating the underlying trend in the data. 
Briefly explain the approach you choose.

#Write a Python function to detect outliers in experimental data.
#Explain the method briefly and implement it using NumPy.
````

````python
%%ai gemini:gemini-2.5-flash



You have a NumPy array containing noisy experimental measurements.
Show a concise Python example for estimating the underlying trend in the data. 
Briefly explain the approach you choose.

#Write a Python function to detect outliers in experimental data.
#Explain the method briefly and implement it using NumPy.
````

````python
%%ai openai-chat:gemma-3-27b-it

A researcher wrote the following Python function to normalize experimental data, but it sometimes returns NaN values.
Explain what the bug is and provide a corrected version of the function. Keep the explanation concise.

import numpy as np

def normalize(x):
    return (x - np.mean(x)) / np.std(x)

data = np.array([5, 5, 5, 5])
print(normalize(data))
````
